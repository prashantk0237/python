```python
principal= float(input("Enter the principal amount: "))
time = float(input('Enter the time year: '))
rate = float(input("Enter the rate" ))
interest_amount= (principal*rate*time)/100
total_amount=principal + interest_amount
print("principal: Rs.", principal)
print("time: ",time ,"year")
print("rate: ", rate, "%")
print("interest_amonut: Rs", interest_amount)
print("total_amount: Rs ",total_amount)
```
